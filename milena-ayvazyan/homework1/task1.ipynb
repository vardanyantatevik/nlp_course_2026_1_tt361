```python
text = """Natural Language Processing (NLP) is a subfield of linguistics, computer science, and artificial
intelligence concerned with the interactions between computers and human language. It's used to
analyze text, allowing machines to understand, interpret, and manipulate human language. NLP has
many real-world applications, including machine translation, sentiment analysis, and chatbots."""
```

```python
# Step 1: Tokenization
# .split() breaks the text into tokens on any whitespace (spaces / newlines).
text = text.split()
text
```

```python
# Step 2: Lowercasing
# Convert every token to lowercase so that e.g. "NLP" and "nlp" are treated as the same word.
text = [i.lower() for i in text]
text
```

```python
# Step 3: Punctuation removal
# .strip("()., '") removes any of these characters ONLY from the start/end of a token
# (e.g. "science," -> "science", "(nlp)" -> "nlp").
# .replace("'s", "") additionally cleans the possessive/contraction "it's" -> "it",
# since the apostrophe there sits in the middle of the token and strip() alone can't reach it.
text = [i.strip("() . ,") for i in text]
text = [i.replace("'s", "") for i in text]
text
```

```python
# Step 4: Stop word removal
# Remove common low-information words using a small hand-picked stop word list
stop_words = ["the", "a", "an", "in", "on", "at", "for", "to", "of",
"and", "is", "are"]
text = [i for i in text if i not in stop_words]
```

```python
# Step 5: Stemming
# Naive rule-based stemmer: strip "ing" or "ed" endings to approximate the word root.
# This is crude on purpose (per the assignment hint) - e.g. "used" -> "us" is a known false stem.
def stemming(word):
  if word.endswith("ing"):
    return word[:-3]
  elif word.endswith("ed"):
    return word[:-2]
  return word

stemmed_tokens = [stemming(i) for i in text]
stemmed_tokens
```

```python
# Bonus: Lemmatization (dictionary-based)
# Maps a handful of common words to their proper base/dictionary form.
# Unlike stemming, this gives linguistically correct roots but only for words listed in the dict.
dct = {
    "is": "be",
    "are": "be",
    "has": "have",
    "computers": "computer",
    "used": "use",
    "machines": "machine",
    "concerned": "concern",
}
def lemmatization(word):
  if word in dct:
    return dct[word]
  return word

lemmatized_tokens = [lemmatization(i) for i in text]
lemmatized_tokens
```

```python
# Comparison of Stemming and Lemmatization results
for stem, lem in zip(stemmed_tokens, lemmatized_tokens):
    print(stem, "-", lem)
```
