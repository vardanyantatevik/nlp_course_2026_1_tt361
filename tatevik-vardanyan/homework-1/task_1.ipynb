```python
import string

#text
text='''Natural Language Processing (NLP) is a subfield of linguistics, computer science, and artificial
intelligence concerned with the interactions between computers and human language. It's used to
analyze text, allowing machines to understand, interpret, and manipulate human language. NLP has
many real-world applications, including machine translation, sentiment analysis, and chatbots.'''
print("\n", "Original text")
print("\n", text)
```

```python
#Tokenization
token = text.split()
print("\n", "Tokenization")
print("\n", token)
```

```python
#Lowercasing
token = [word.lower() for word in token]
print("\n", "Lowercasing")
print("\n", token)
```

```python
#Punctuation removal
token = [word.translate(str.maketrans('', '', string.punctuation)) for word in token]
token = [word for word in token if word]
print("\n", "Punctuation removal")
print("\n", token)
```

```python
#Stop word removal
stop = {
    "the", "a", "an", "in", "on", "at", "for", "to", "of", "and", "is", "are"
}
token = [word for word in token if word not in stop]
print("\n", "Stop word removal")
print("\n", token)
```

```python
#Stemming
def stemming(word):
    suffixs = ["ing", "ed", "ly", "es", "s"]

    for suffix in suffixs:
        if word.endswith(suffix) and len(word) > len(suffix) + 3 :
            return word[:-len(suffix)]
    return(word)
tokens = [stemming(word) for word in token]

print("\n", "Stemming")
print("\n", tokens)
```

```python
print("\n", "Final text")
print("\n", *tokens)
```
